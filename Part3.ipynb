```python
def classify_review(review):
    count=0
    c_count = sum(review.count(word) for word in c_keywords)
    b_count = sum(review.count(word) for word in b_keywords)
    h_count = sum(review.count(word) for word in h_keywords)

    counts = {'c': c_count, 'b': b_count, 'h': h_count}
    
    max_category = max(counts, key=counts.get)
    
    if counts[max_category] == 0:
        return 'NOT KNOWN'
    
    return max_category

df_test['predicted_label'] = df_test['text'].apply(classify_review)
```

# Domain Transfer of SVM

```python
import os
import jieba
import pandas as pd
import numpy as np
from tqdm import tqdm
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import f1_score
```

```python
BASE_DIR = r"C:\Users\fayec\Desktop\Master\Master\7404 Computational intelligence and machine learning\group project"
TRAIN_DIR = os.path.join(BASE_DIR, "training set")
TEST_DIR = os.path.join(BASE_DIR, "testing set")


DOMAINS = {
    "ChnSentiCorpEdu": "ChnSentiCorpEdu.csv",
    "ChnSentiCorpMov": "ChnSentiCorpMov.csv",
    "ChnSentiCorpHou": "ChnSentiCorpHou.csv"
}
```

```python
def load_dataset(file_path):
    df = pd.read_csv(file_path)  # 读取 CSV 文件
    df.dropna(inplace=True)  # 移除空值
    df["text"] = df["text"].astype(str).apply(lambda x: " ".join(jieba.cut(x)))  # 进行分词
    return df
```

```python
train_data, test_data = {}, {}

print("Loading Training Dataset...")
for domain, filename in tqdm(DOMAINS.items()):
    train_data[domain] = load_dataset(os.path.join(TRAIN_DIR, filename))

print("Loading Testing Dataset...")
for domain, filename in tqdm(DOMAINS.items()):
    test_data[domain] = load_dataset(os.path.join(TEST_DIR, filename))
```

```python
print("TF-IDF vectoring...")
vectorizer = TfidfVectorizer(max_features=10000)  # 6000 features
all_texts = pd.concat([df["text"] for df in train_data.values()] + [df["text"] for df in test_data.values()])
vectorizer.fit(all_texts)  # Train TF-IDF word vector
```

```python
results = {}

print("Train SVM and Transfer domain test...")
for train_domain in tqdm(DOMAINS.keys(), desc="Train SVM"):
    X_train = vectorizer.transform(train_data[train_domain]["text"])
    y_train = train_data[train_domain]["label"]

    # Train SVM model
    model = SVC(kernel="linear")
    model.fit(X_train, y_train)

    for test_domain in DOMAINS.keys():
        if train_domain == test_domain:
            continue 

        # Test data
        X_test = vectorizer.transform(test_data[test_domain]["text"])
        y_test = test_data[test_domain]["label"]
        y_pred = model.predict(X_test)

        # Calculate Micro-F1 and Macro-F1
        micro_f1 = f1_score(y_test, y_pred, average="micro")
        macro_f1 = f1_score(y_test, y_pred, average="macro")

        # Record the result
        results[(train_domain, test_domain)] = {"Micro-F1": micro_f1, "Macro-F1": macro_f1}
```

```python
df_results = pd.DataFrame.from_dict(results, orient="index")
df_results.index.names = ["Training Set", "Testing Set"]

print("\nCross-domain Sentiment Analysis Results:")
print(df_results)
```
